# Token estimate — embedding `joint_dataset` with voyage-4

Rough estimate of how many tokens it takes to embed all nodes of `joint_dataset/nodes.csv`.

Text per node = `title + "\n\n" + abstract`.

Token count is approximated as `words * 1.3` (typical for English prose), not the real tokenizer.

```python
import pandas as pd

TOKENS_PER_WORD = 1.3   # rough approximation
PRICE_PER_1M = 0.06     # USD per 1M tokens — verify at https://docs.voyageai.com/docs/pricing

nodes = pd.read_csv("../joint_dataset/nodes.csv", usecols=["title", "abstract"])
texts = (nodes["title"].fillna("") + "\n\n" + nodes["abstract"].fillna("")).tolist()
print(f"{len(texts):,} nodes")
```

```python
words = sum(len(t.split()) for t in texts)
total = round(words * TOKENS_PER_WORD)
```

```python
print(f"total words      : {words:,}")
print(f"total tokens     : {total:,}")
print(f"mean per node    : {total / len(texts):,.0f}")
print(f"cost at ${PRICE_PER_1M}/1M : ${total / 1e6 * PRICE_PER_1M:,.2f}")
```
